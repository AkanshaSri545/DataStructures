2.How does broadcasting work in NumPy?
Broadcasting in NumPy is a powerful mechanism that allows array operations between arrays of different shapes without the need to manually replicate data. It lets NumPy perform element-wise operations efficiently on arrays that don't have the exact same shape, by "stretching" the smaller array(s) across the larger ones, where possible.

🔧 How Broadcasting Works
When performing operations on two arrays (like addition or multiplication), NumPy compares their shapes element-wise from the right. The rules for broadcasting are:

If the arrays have different numbers of dimensions, the shape of the smaller array is padded with ones on the left until both arrays have the same number of dimensions.

Arrays are compatible in a dimension if:

They are equal in that dimension, or

One of them is 1 (this dimension will be "broadcast" to match the other).

8. What does the term “vectorized operation” mean in NumPy?
In NumPy, the term "vectorized operation" refers to the ability to perform element-wise operations on entire arrays (or large chunks of data) without the need for explicit loops. Instead of iterating over each element manually, NumPy allows operations to be applied to entire arrays at once, which significantly improves both code simplicity and execution speed. This is made possible because NumPy operations are implemented in highly optimized, low-level C code, which can handle array operations much more efficiently than Python’s native looping constructs.

For example, when adding two arrays together, you don't need to loop through each element individually. NumPy internally takes care of iterating through the arrays and applying the operation element-wise in a single step, providing a more concise and faster alternative. Vectorized operations take full advantage of SIMD (Single Instruction, Multiple Data) instructions in modern CPUs, enabling them to perform calculations on multiple data points simultaneously, further speeding up execution.

21. Why is NumPy’s array slicing faster than Python’s list slicing?
**NumPy's array slicing** is faster than **Python's list slicing** because of the way NumPy arrays are implemented at a lower level, making them more efficient for large-scale operations. Here’s why:

1. **Memory Layout and Contiguity**:
   - NumPy arrays are stored in contiguous memory blocks, meaning that the elements of an array are stored in adjacent memory locations. This allows NumPy to perform slicing operations very efficiently because it can directly access the memory addresses of the sliced elements, without needing to iterate through the data structure or create additional copies.
   - In contrast, Python lists are arrays of pointers to objects, which can lead to fragmentation in memory. This fragmentation can result in slower performance when accessing or slicing elements because Python lists require more memory overhead for managing their dynamic size and underlying data types.

2. **View vs. Copy**:
   - NumPy slicing generally returns a **view** of the original array, meaning that it does not copy the data but rather creates a new array that references the original data in memory. This means that slicing a NumPy array is a lightweight operation, as no new data needs to be created.
   - On the other hand, Python lists create a **new copy** of the sliced portion, which involves additional memory allocation and copying of elements. This copying process takes time and results in slower performance compared to NumPy’s approach.

3. **Vectorization and Optimized Operations**:
   - NumPy is designed to take advantage of **vectorization**, which allows it to perform operations on entire arrays at once using highly optimized C and Fortran libraries under the hood. This enables NumPy to execute slicing and subsequent operations in a single step, minimizing overhead.
   - Python lists, being more general-purpose, lack such low-level optimization and require looping or more complex logic for operations like slicing, making them slower for large data manipulations.

4. **No Type Checking at Runtime**:
   - NumPy arrays are homogeneous, meaning that they store elements of a single data type (e.g., all integers or all floats). This allows NumPy to optimize memory access and operations, as it doesn't need to check or handle different types of data at runtime, unlike Python lists, which can store elements of different types and require additional checks during operations.

In summary, NumPy’s array slicing is faster than Python’s list slicing because NumPy arrays are stored in contiguous memory, return views instead of copies, benefit from vectorization, and avoid type checking overhead. This makes NumPy more efficient for handling large datasets and performing slicing operations compared to Python lists.

```python
#2 Write a Pandas script to find the mean of a specific column in a DataFrame.

import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'Salary': [50000, 55000, 60000, 65000, 70000]
}

df = pd.DataFrame(data)

mean_age = df['Age'].mean()

print("Mean of Age column:", mean_age)
```

```python
#3 Create a scatter plot using Matplotlib.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 7, 9, 11, 13]

plt.scatter(x, y, color='blue', label='Data Points')

plt.title('Simple Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.legend()

plt.show()
```

```python
#4 How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'A': [1, 2, 3, 4, 5],
    'B': [5, 6, 7, 8, 9],
    'C': [9, 8, 7, 6, 5]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')

plt.show()
```

```python
#5 Generate a bar plot using Plotly.

import plotly.express as px

data = {'Category': ['A', 'B', 'C', 'D'], 'Values': [10, 20, 30, 40]}

df = pd.DataFrame(data)

fig = px.bar(df, x='Category', y='Values', title='Bar Plot')
fig.show()
```

```python
#8 Create a line plot with multiple lines using Matplotlib.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [5, 4, 3, 2, 1]

plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')

plt.legend()
plt.show()
```

```python
#9 Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.

import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40]}

df = pd.DataFrame(data)

filtered_df = df[df['Age'] > 30]

filtered_df
```
